### Representing 32-bit words in Python

A 32-bit word contains 32 bits (4 bytes), giving $2^{32}$ possible bit patterns. Interpreted as an unsigned integer, its value ranges from 0 to $2^{32} - 1$. SHA-256 uses 32-bit words as part of its algorithm definition ([Secure Hash Standard](https://nvlpubs.nist.gov/nistpubs/FIPS/NIST.FIPS.180-4.pdf)).

In Python, one way to represent such a value is with NumPy’s `uint32` type ([NumPy documentation](https://numpy.org/doc/stable/reference/arrays.scalars.html)).

```python
# Create a NumPy unsigned 32-bit integer scalar
word = np.uint32(0x12345678)
print(word)
type(word)
```
